```python
from dotenv import load_dotenv

load_dotenv()
```

```python
from dataclasses import dataclass

@dataclass
class ColourContext:
    favorite_colour: str = "blue"
    least_favorite_colour: str = "yellow"
```

```python
from langchain.agents import create_agent

agent = create_agent(
    model="ollama:qwen2.5:7b",
    context_schema=ColourContext  
)
```

```python
from langchain.messages import HumanMessage

response = agent.invoke(
    {"messages": [HumanMessage(content="What is my favorite colour?")]},
    context=ColourContext()
)
```

```python
from pprint import pprint

pprint(response)
```

## Accessing Context

```python
from langchain.tools import tool, ToolRuntime

@tool
def get_favorite_colour(runtime: ToolRuntime) -> str:
    """Get the favorite colour of the user"""
    return runtime.context.favorite_colour

@tool
def get_least_favorite_colour(runtime: ToolRuntime) -> str:
    """Get the least favorite colour of the user"""
    return runtime.context.least_favorite_colour
```

```python
agent = create_agent(
    model="ollama:qwen2.5:7b",
    tools=[get_favorite_colour, get_least_favorite_colour],
    context_schema=ColourContext
)
```

```python
response = agent.invoke(
    {"messages": [HumanMessage(content="What is my favorite colour?")]},
    context=ColourContext()
)

pprint(response)
```

```python
response = agent.invoke(
    {"messages": [HumanMessage(content="What is my favorite colour?")]},
    context=ColourContext(favorite_colour="green")
)

pprint(response)
```
